```python
X=pd.DataFrame({
    'col_1':[1,2,3,None,None,100,12,95,45,25],
    'col_2':[101,102,145,None,104,None,189,190,None,200],
    'col_3':['A','B','C',None,'A','B','A','C',None,'C'],
    'col_4':['AA','AB','CC',None,'None','BB',None,'CC',None,'CC']
})
X
```

```python
X['col_3']=X['col_3'].astype(str)
X['col_4']=X['col_4'].astype(str)
```

```python
imputer=SimpleImputer(strategy="constant",fill_value='unknown')
imputer.fit_transform(X[['col_3','col_4']])
```
